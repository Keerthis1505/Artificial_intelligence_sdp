```python
# Basic data types in Python

# Integers
age = 25
count = -10

# Floats
price = 19.99
temperature = -5.5

# Strings
name = "Alice"
message = 'Hello, World!'

# Booleans
is_active = True
is_empty = False

# Lists
numbers = [1, 2, 3, 4, 5]
mixed_list = [1, "two", 3.0, True]

# Tuples
coordinates = (10, 20)
rgb = (255, 128, 0)

# Dictionaries
person = {"name": "Bob", "age": 30, "city": "New York"}
settings = {"theme": "dark", "volume": 80}

# Sets
unique_numbers = {1, 2, 3, 4, 5}
letters = {"a", "b", "c"}

# None type
result = None
```

```python
if age > 18:
    print(f"{name} is an adult.")
else:
    print(f"{name} is not an adult.")
```

```python
variables = [
    ("age", age),
    ("count", count),
    ("price", price),
    ("temperature", temperature),
    ("name", name),
    ("message", message),
    ("is_active", is_active),
    ("is_empty", is_empty),
    ("numbers", numbers),
    ("mixed_list", mixed_list),
    ("coordinates", coordinates),
    ("rgb", rgb),
    ("person", person),
    ("settings", settings),
    ("unique_numbers", unique_numbers),
    ("letters", letters),
    ("result", result),
]

for var_name, value in variables:
    print(f"{var_name}: {value} ({type(value).__name__})")
```

```python
if price > 20:
    print("The price is high.")
else:
    print("The price is affordable.")
```
